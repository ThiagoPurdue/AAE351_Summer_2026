# 4. Functional Decomposition and Concept Synthesis

## 4.1 Functional Decomposition
Before deciding *how* a system will be built (the physical architecture), engineers must define *what* the system must do (the functional architecture). Functional decomposition breaks down the top-level mission (e.g., "Transport Cargo") into sub-functions.

> **💡 Instructor's Pro-Tip:** 
> *"A common mistake is assigning physical parts during functional analysis. Do not write 'Propeller' in your functional diagram! A propeller is a physical solution. The function is 'Provide Thrust'. If you lock in 'Propeller' too early, you blind yourself to other solutions like jet engines or ducted fans."*

## 4.2 Morphological Analysis
With functions defined, we explore the design space using a **Morphological Matrix**. The team lists all primary functions on the vertical axis and brainstorms every possible physical mechanism that could satisfy that function on the horizontal axis. 

![Morphological Matrix](../images/05_Concept_Generation_and_Selection_p10_i12.png)
*Figure 4.1: Morphological Matrix for concept synthesis applied to structural and propulsion configurations. Drawing lines through the matrix creates unique vehicle architectures.*

> **📘 Industry Insight: INCOSE Systems Engineering Vision**
> INCOSE strongly advocates for Model-Based Systems Engineering (MBSE) during Concept Synthesis. By using tools like MATLAB System Composer, engineers can evaluate thousands of morphological combinations through automated simulations rather than relying on manual, intuition-based down-selection.

## 4.3 Concept Synthesis
Drawing paths through the morphological matrix generates distinct concepts. These concepts are then subjected to initial feasibility screening to eliminate those that violate physical constraints before moving to formal evaluation via a Pugh Matrix.

## Applying this to the RCUAV Project
When mapping functions to physical forms for the 100% 3D-printed RCUAV, special attention must be paid to the propulsion/structure interface. 
**Functional Decomposition for Propulsion Interface:**
* **Function 1:** Provide Thrust $\rightarrow$ *Physical Solution:* Brushless DC Motor
* **Function 2:** Transfer Load to Airframe $\rightarrow$ *Physical Solution:* 3D-Printed Motor Mount (100% Infill PETG)
* **Function 3:** Dissipate Heat $\rightarrow$ *Physical Solution:* Integrated NACA cooling ducts in the printed cowling.

Notice how the physical solutions are specifically tailored to the Additive Manufacturing constraint.\n

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Generate a Functional Block Diagram for the RCUAV Interface
G = nx.DiGraph()

# Define functions
edges = [
    ("Provide Thrust", "Transfer Load to Airframe"),
    ("Provide Thrust", "Dissipate Heat"),
    ("Transfer Load to Airframe", "Maintain Structural Integrity"),
    ("Dissipate Heat", "Maintain Operating Temp")
]

G.add_edges_from(edges)

# Define a custom layout for a block diagram feel
pos = {
    "Provide Thrust": (0, 1),
    "Transfer Load to Airframe": (1, 1.5),
    "Dissipate Heat": (1, 0.5),
    "Maintain Structural Integrity": (2, 1.5),
    "Maintain Operating Temp": (2, 0.5)
}

plt.figure(figsize=(10, 5))
nx.draw_networkx_nodes(G, pos, node_color='lightgreen', node_shape='s', node_size=4000)
nx.draw_networkx_edges(G, pos, arrowstyle='->', arrowsize=20, edge_color='gray')
nx.draw_networkx_labels(G, pos, font_size=9, font_weight='bold', 
                        labels={n: n.replace(' ', '\n') for n in G.nodes()})

plt.title("Functional Flow: Propulsion & Structure Interface", fontsize=14)
plt.axis('off')
plt.show()\n
```
