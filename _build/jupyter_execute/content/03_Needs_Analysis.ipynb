# 3. Needs Analysis and Objective Trees

## 3.1 Understanding the Stakeholder
Before a single line of code is written or a CAD sketch is drawn, engineers must understand the fundamental problem they are trying to solve. This begins with **Stakeholder Elicitation**. Stakeholders are not just the immediate customers purchasing the aircraft; they include end-users, maintainers, regulatory bodies, and manufacturers. Gathering these needs often involves interviews, market research, and analyzing legacy systems.

> **💡 Instructor's Pro-Tip:** 
> *“I always tell my students: the customer rarely knows what they actually want, but they always know what they don’t want. Your job is to listen to their frustrations with current systems and translate those into positive, measurable engineering needs. Don't start designing a wing when you haven't even verified they need to fly!”*

## 3.2 Hierarchical Objective Trees
To make sense of disparate and often conflicting stakeholder needs, we organize them into **Hierarchical Objective Trees**. This visual breakdown ensures that high-level abstract goals (e.g., "The vehicle must be safe") are decomposed into lower-level, actionable objectives (e.g., "The vehicle must survive a single motor failure").

![Objective Tree](../images/03_Needs_Analysis_p10_i17.png)
*Figure 3.1: Sample Objective Tree representing stakeholder needs mapping for the RCUAV project. Notice how top-level abstract needs decompose into specific sub-objectives.*

> **📘 Industry Insight: NASA Systems Engineering Handbook**
> According to the NASA SE Handbook (Rev 2), Section 4.1, Stakeholder Expectations Definition is the first step in the Systems Engineering Engine. It emphasizes that poorly defined expectations inevitably lead to requirements churn, schedule overruns, and ultimately, mission failure. INCOSE standards echo this, requiring formal baseline reviews of needs before requirement formulation begins.

## 3.3 Translating Needs to Requirements
The final step of needs analysis is transitioning from qualitative desires to quantitative metrics. If a stakeholder needs the aircraft to be "fast," the systems engineer must frame this as a specific cruise velocity requirement.

## Applying this to the RCUAV Project
For the **Regional Cargo UAV (RCUAV)**, your objective tree must account for the 100% 3D-printed constraint. Based on our ConOps, here are 5 sample requirements derived from the initial needs analysis:

1. **Payload Requirement:** The RCUAV shall carry a minimum payload of 2.5 kg.
2. **Range Requirement:** The RCUAV shall possess a minimum operational range of 15 km on a single battery charge.
3. **Manufacturability Requirement:** The primary airframe structure shall be manufacturable exclusively using FDM 3D printing in PLA+ or PETG.
4. **Assembly Requirement:** The airframe components shall assemble without the use of chemical adhesives (e.g., using printed snap-fits or standard fasteners).
5. **Reliability Requirement:** The RCUAV shall maintain structural integrity under a 3G positive load factor.\n

```python
import networkx as nx
import matplotlib.pyplot as plt

# Generate a visual Objective Tree for the RCUAV
G = nx.DiGraph()

# Define nodes
nodes = [
    ("Safe Operation", "RCUAV Objectives"),
    ("Deliver Cargo", "RCUAV Objectives"),
    ("Manufacturability", "RCUAV Objectives"),
    ("Redundant Propulsion", "Safe Operation"),
    ("Structural Integrity (3G)", "Safe Operation"),
    ("2.5 kg Payload", "Deliver Cargo"),
    ("15 km Range", "Deliver Cargo"),
    ("FDM 3D Printable", "Manufacturability"),
    ("No Chemical Adhesives", "Manufacturability")
]

G.add_edges_from(nodes)

# Plotting the tree
plt.figure(figsize=(12, 6))
pos = nx.drawing.nx_agraph.graphviz_layout(G, prog='dot') if hasattr(nx.drawing, 'nx_agraph') else nx.spring_layout(G)
# Using a fallback layout if graphviz is not installed
pos = nx.spring_layout(G, k=0.9, iterations=50)

nx.draw(G, pos, with_labels=True, node_color='lightblue', 
        node_size=3000, font_size=10, font_weight='bold', arrows=False)

plt.title("RCUAV Hierarchical Objective Tree", fontsize=14)
plt.show()\n
```
